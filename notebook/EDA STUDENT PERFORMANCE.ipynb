#### 4.2 Maximumum score of students in all three subjects

A violin plot shows the distribution of your data.

A violin plot helps you understand:
- where most scores are concentrated
- where scores are less common
- how spread out the scores are
- the shape of the distribution
- possible extreme values

The width of the violin represents the density of observations.

```python
plt.figure(figsize=(18,8))
plt.subplot(1,3,1)
plt.title('Maths Scores')
sns.violinplot(data=df,y='math_score',color='red',linewidth=3) #Notice there is no x.That's because this is a one-dimensional distribution.
plt.subplot(1,3,2)
plt.title('Reading Scores')#linewidth ->This controls the thickness of the violin plot's outline.Higher value → thicker outline.
sns.violinplot(data=df,y='reading_score',color='green',linewidth=3) 
#Wider = more observations/density around that score.Narrower = fewer observations/density around that score.
plt.subplot(1,3,3)
plt.title('Writing Scores')
sns.violinplot(data=df,y='writing_score',color='blue',linewidth=3)
plt.show()
```

#### Insights
- From the above three plots its clearly visible that most of the students score in between 60-80 in Maths whereas in reading and writing most of them score from 50-80

#### 4.3 Multivariate analysis using pieplot

- plt.rcParams contains Matplotlib's default settings.
- size-> contains the number of students in each category.
- autopct-> Automatically put the percentage on each pie slice.
- tight_layout-> This is used to automatically adjust spacing between your subplots.Without it, your five graphs/titles/labels might overlap.


```python
##Setting the figure size
plt.rcParams['figure.figsize']=(30,12) #Here you're changing the default figure size to:width=30, height=12

##first subplot
plt.subplot(1,5,1)
#Counting genders
size=df['gender'].value_counts()#A pie chart needs numbers to determine the size of each slice.
#Creating labels
labels='Male','Female'#This creates the names that will appear next to the pie slices.
#Creating colors
color=['green','red']
#Creating the pie chart
plt.pie(size,colors=color,labels=labels,autopct='%.2f%%')
#Title
plt.title('Gender',fontsize=20)
#Turning off the axes
plt.axis('off')#pie chart doesn't need X/Y axes

##second subplot
plt.subplot(1,5,2)
size=df['race_ethnicity'].value_counts()
labels='Group C', 'Group D','Group B','Group E','Group A'
color = ['red', 'green', 'blue', 'cyan','orange']
plt.pie(size,colors=color,labels=labels,autopct='%.2f%%')
plt.title('Race Ethnicity',fontsize=20)
plt.axis('off')

##third subplot
plt.subplot(1,5,3)
size=df['lunch'].value_counts()
labels='Standard', 'Free'
color=['red','green']
plt.pie(size,colors=color,labels=labels,autopct='%.2f%%')
plt.title('Lunch',fontsize=20)
plt.axis('off')

##fourth subplot
plt.subplot(1,5,4)
size = df['test_preparation_course'].value_counts()
labels = 'None', 'Completed'
color = ['red','green']
plt.pie(size, colors = color,labels = labels,autopct = '.%2f%%')
plt.title('Test Course', fontsize = 20)
plt.axis('off')

##fifth subplot
plt.subplot(1, 5, 5)
size = df['parental_level_of_education'].value_counts()
labels = 'Some College', "Associate's Degree",'High School','Some High School',"Bachelor's Degree","Master's Degree"
color = ['red', 'green', 'blue', 'cyan','orange','grey']
plt.pie(size, colors = color,labels = labels,autopct = '.%2f%%')
plt.title('Parental Education', fontsize = 20)
plt.axis('off')

plt.tight_layout() #Arrange everything so that it fits nicely.
plt.grid() #not might be useful here because we have made axes off
plt.show()
```

#####  Insights
- Number of Male and Female students is almost equal
- Number students are greatest in Group C
- Number of students who have standard lunch are greater
- Number of students who have not enrolled in any test preparation course is greater
- Number of students whose parental education is "Some College" is greater followed closely by "Associate's Degree"

#### 4.4.1 GENDER COLUMN
- How is distribution of Gender ?
- Is gender has any impact on student's performance ?

#### UNIVARIATE ANALYSIS ( How is distribution of Gender ? )

- A count plot essentially answers:How many observations are there in each category?
- controls how intense/saturated the colors are.
  1 = full saturation.
  0 = completely desaturated.
So 0.95 means the colors are almost fully saturated.
- .containers->A count plot contains graphical objects representing the bars.
- explode->This controls whether a slice is pulled away from the center


```python
f,ax=plt.subplots(1,2,figsize=(20,10))
sns.countplot(data=df,x=df['gender'],palette='bright',ax=ax[0],saturation=0.95)

for container in ax[0].containers:#ax[0].containers allows us to access the bar containers.
    ax[0].bar_label(container,color='black',size=20)#This adds the actual count number on top of each bar.

ax[1].pie(x=df['gender'].value_counts(),labels=['Female','Male'],explode=[0,0.1],autopct='%1.1f%%',shadow=True,
          colors=['#ff4d4d','#ff8000'])
plt.show()
```

#### Insights 
- Gender has balanced data with female students are 518 (52%) and male students are 482 (48%) 

#### BIVARIATE ANALYSIS ( Is gender has any impact on student's performance ? ) 

```python
gender_group=df.groupby('gender').mean(numeric_only=True)
gender_group
```

```python
#male_scores->These are the heights of the bars.
#width->0.4
plt.figure(figsize=(10,8))
X=['Total Average','Math Average']
female_scores=[
    gender_group.loc['female','Average'],
    gender_group.loc['female','math_score']
]
male_scores=[
    gender_group.loc['male','Average'],
    gender_group.loc['male','math_score']
]
#Create positions for the categories
X_axis=np.arange(len(X))#X_axis=[0,1]

plt.bar(X_axis-0.2,male_scores,0.4,label='Male') 
plt.bar(X_axis+0.2,female_scores,0.4,label='Female')

plt.xticks(X_axis,X) #X_axis = [0, 1],X = ['Total Average', 'Math Average']->This tells Matplotlib:At position 0, write "Total Average" 
#at position 1, write "Math Average"
plt.ylabel("Marks")
plt.title("Total average v/s Math average marks of both the genders", fontweight='bold')
plt.legend()
plt.show()
```

#### Insights 
- On an average females have a better overall score than men.
- whereas males have scored higher in Maths.

#### UNIVARIATE ANALYSIS ( How is Group wise distribution ?)

```python
f,ax=plt.subplots(1,2,figsize=(20,10))
sns.countplot(data=df,x=df['race_ethnicity'],ax=ax[0],palette='bright',saturation=0.95)
for container in ax[0].containers:
    ax[0].bar_label(container,color='black',size=20)
ax[1].pie(x=df['race_ethnicity'].value_counts(),labels=df['race_ethnicity'].value_counts().index,shadow=True,explode=[0.1,0,0,0,0],
          autopct='%1.1f%%')
plt.show()
```

#### BIVARIATE ANALYSIS ( Is Race/Ehnicity has any impact on student's performance ? )

```python
# Group the data by race_ethnicity
Group_data2 = df.groupby('race_ethnicity')

# Calculate average scores for each group
math_avg = Group_data2['math_score'].mean()
reading_avg = Group_data2['reading_score'].mean()
writing_avg = Group_data2['writing_score'].mean()

# Create the figure
plt.figure(figsize=(20, 8))

# Get the race/ethnicity categories
X = math_avg.index

# Create positions for the groups
X_axis = np.arange(len(X))

# Set bar width
width = 0.25

# Create bars
plt.bar(X_axis - width, math_avg.values, width, label='Math')
plt.bar(X_axis, reading_avg.values, width, label='Reading')
plt.bar(X_axis + width, writing_avg.values, width, label='Writing')

# Add labels on X-axis
plt.xticks(X_axis, X)

# Add axis labels
plt.xlabel('Race_Ethnicity')
plt.ylabel('Average Marks')

# Add title
plt.title(
    'Average Math, Reading and Writing Scores by Race_Ethnicity',
    fontweight='bold'
)

# Add legend
plt.legend()

# Display the graph
plt.show()
```

#### Insights 
- Group E students have scored the highest marks. 
- Group A students have scored the lowest marks. 
- Students from a lower Socioeconomic status have a lower avg in all course subjects

#### 4.4.7 MUTIVARIATE ANALYSIS USING PAIRPLOT

A pairplot creates multiple plots showing the relationship between every numerical variable and every other numerical variable in your DataFrame.

This is called bivariate analysis for each individual pair, while the complete collection of relationships is being used for multivariate analysis.

hue(gender)-> Use gender to differentiate the observations.

Take all the numerical variables in my DataFrame, compare every variable with every other variable, show the distributions on the diagonal, and use different colors for male and female students.

It helps us quickly identify:
- Relationships between variables
- Positive/negative correlations
- Patterns
- Clusters
- Possible outliers
- Differences between groups such as gender
- Distributions of individual numerical variables

```python
sns.pairplot(df,hue = 'gender')
plt.show()
```
